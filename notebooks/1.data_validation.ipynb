````python
import polars as pl
import boto3
import mlflow
import os
from datetime import datetime
from dotenv import load_dotenv
load_dotenv()
````

````python
aws_access_key_id = os.getenv("AWS_ACCESS_KEY_ID")
aws_secret_access_key = os.getenv("AWS_SECRET_ACCESS_KEY")
endpoint_url = os.getenv("MLFLOW_S3_ENDPOINT_URL")
bucket_name = os.getenv("BUCKET_NAME")
bucket_key = os.getenv("BUCKET_KEY")

s3 = boto3.client(
    "s3",
    aws_access_key_id=aws_access_key_id,
    aws_secret_access_key=aws_secret_access_key,
    endpoint_url=endpoint_url
)
bucket_key
````

````python
obj=s3.get_object(Bucket=bucket_name, Key=bucket_key)
obj
````

````python
df = pl.read_csv(obj['Body'])
len(df)
````

````python
report = {
        "status": "PASSED",
        "errors": [],
        "warnings": [],
        "stats": {},
    }
````

Column check

````python
required_columns = ['Brand', 'Quantity Sold', 'Dispatch Date','Region']
missing_columns = [col for col in required_columns if col not in df.columns]
if missing_columns:
    report["status"] = "FAILED"
    report["errors"].append(f"Missing required columns: {', '.join(missing_columns)}")
report
````

Data type check

````python
# Ensure 'Date' is parseable as date
try:
    df = df.with_columns([
        pl.col("Dispatch Date").str.strptime(pl.Date, format="%Y-%m-%d"),
        pl.col("Inward Date").str.strptime(pl.Date, format="%Y-%m-%d")
    ])
    report["stats"]["date_conversion"] = "OK"
except Exception as e:
    report["status"] = "FAILED"
    report["stats"]["date_conversion"] = "FAILED"
    report["errors"].append(f"Error occurred while converting dates: {e}")
report
````

````python
# Ensure 'Quantity Sold' is numeric

if df['Quantity Sold'].dtype not in [pl.Int64]:
    report["status"] = "FAILED"
    report["errors"].append("'Quantity Sold' is not numeric")
    report["stats"]["quantity_sold_type"] = "FAILED"
report
````

Null Value Checks

````python
null_counts = df.null_count()

null_columns = [col for col in df.columns if null_counts[col][0] > 0]
if null_columns:
    report["status"] = "FAILED"
    report["errors"].append(f"Columns with null values: {', '.join(null_columns)}")
    report["stats"]["null_columns"] = {col: null_counts[col][0] for col in null_columns}
report
````

Range & Value Constraints

````python
# Quantity Sold must be >= 0
df["Quantity Sold"]
if df["Quantity Sold"].min() < 0:
    report["status"] = "FAILED"
    report["errors"].append("'Quantity Sold' contains negative values")
    report["stats"]["quantity_sold_min"] = df["Quantity Sold"].min()
report
````

````python
# Price must be >= 0
if df["Price"].min() < 0:
    report["status"] = "FAILED"
    report["errors"].append("'Price' contains negative values")
    report["stats"]["price_min"] = df["Price"].min()
report
````

Duplicate Rows

````python
duplicates = df.is_duplicated()
if duplicates.any():
    report["status"] = "FAILED"
    report["errors"].append("Duplicate rows found")
    report["stats"]["duplicate_rows"] = duplicates.sum()
report
````

Data Freshness

````python
date_max = df["Dispatch Date"].cast(pl.String).str.strptime(pl.Date, format="%Y-%m-%d").max()
date_max
````

````python
today = datetime.now().date()
today
````

````python
if date_max > today:
    report["status"] = "FAILED"
    report["errors"].append(f"'Dispatch Date' contains future dates: {date_max}")
    report["stats"]["dispatch_date_max"] = date_max
````
